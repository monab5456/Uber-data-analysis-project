# Data Cleaning and Transformation

```python
print(df['Cancelled Rides by Customer'].value_counts())
print("****"*50)
print(df['Reason for cancelling by Customer'].value_counts())
```

```python
# filling the null values with 0
df['Cancelled Rides by Customer'] = df['Cancelled Rides by Driver'].fillna(0)
```

```python
df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].fillna('Not Mentaioned')
```

```python
print(df['Cancelled Rides by Driver'].value_counts())
print("***"*50)
print(df['Driver Cancellation Reason'].value_counts())
```

```python
df['Cancelled Rides by Driver'] = df['Cancelled Rides by Driver'].fillna(0)
```

```python
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].fillna('Not Mentaioned')
```

```python
print(df['Incomplete Rides'].value_counts())
print("***"*50)
print(df['Incomplete Rides Reason'].value_counts())
```

```python
df['Incomplete Rides'] = df['Incomplete Rides'].fillna(0)
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].fillna('Other Issue')
```

```python
df[(df['Booking Value'].isnull()) | (df['Booking Value'] == 0)].sample(5)
```

```python
df['Booking Value'] = df['Booking Value'].fillna(0)
df['Ride Distance'] = df['Ride Distance'].fillna(0)
```

Keeping the Driver and Customer ratring as null as the ride never happens so imputing the values with '0' wuldn't be appropitate.

```python
df['Payment Method'] = df['Payment Method'].fillna('Not Applicable')
```

**Datatype Transformation**

```python
#changing the datatype to integer
df['Cancelled Rides by Driver'] = df['Cancelled Rides by Driver'].astype(int)
df['Incomplete Rides'] = df['Incomplete Rides'].astype(int)
df['Cancelled Rides by Customer'] = df['Cancelled Rides by Customer'].astype(int)

# Changing the datatype of time column to datetime datatype
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S',errors='coerce').dt.time

# Combine Date and Time into a single DateTime column
df['DateTime'] = df.apply(lambda row: pd.Timestamp.combine(row['Date'].date(), row['Time']), axis=1)
```

# EDA

```python
#Visulising the booking status count
plt.figure(figsize=(9, 6))
sns.countplot(x=df['Booking Status'])
plt.title('Booking Status Count')
plt.tight_layout()
plt.show()
```

```python
#Count of booking by Vechical Type
plt.figure(figsize=(10, 6))
sns.countplot(data=df,x='Vehicle Type', hue='Vehicle Type', palette='viridis', legend=False)
plt.title('Booking Count by Vehicle Type')
plt.tight_layout()
plt.show()
```

```python
# Booking values distribution across different Vehicle type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Vehicle Type', y='Booking Value', hue='Vehicle Type', palette='viridis', legend=False)
plt.title('Booking Value Distribution by Vehicle Type')
plt.tight_layout()
plt.show()
```

```python
# Correlation matrix of numerical columns
cols = ['Booking Value', 'Ride Distance', 'Customer Rating', 'Driver Ratings',
        'Cancelled Rides by Customer', 'Cancelled Rides by Driver', 'Incomplete Rides']

corr_matrix = df[cols].corr()
corr_matrix
```

```python
# Corelation analysis of numerical columns
cols = ['Booking Value', 'Ride Distance', 'Customer Rating', 'Driver Ratings',
            'Cancelled Rides by Customer', 'Cancelled Rides by Driver', 'Incomplete Rides']
plt.figure(figsize=(10, 8))
sns.heatmap(df[cols].corr(), annot=True, cmap='coolwarm', fmt='.2f',linewidths=0.5)
plt.title('Correlation Heatmap')
```

```python
#booking per week per hour
df['week_day'] = df['DateTime'].dt.day_name()
df['hour'] = df['DateTime'].dt.hour

week = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

#pivoting the data - fixing the index key from 'week' to 'week_day'
pivot = df.pivot_table(index='week_day', columns='hour', values='Booking ID', aggfunc='count', fill_value=0)

# Reordering the index to match week order
pivot = pivot.reindex(week)

#plotting the heatmap
plt.figure(figsize=(15, 8))
sns.heatmap(pivot, annot=True, cmap='YlGnBu', fmt='d')
plt.title('Booking Count by Week and Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Day of the Week')
plt.show()
```
